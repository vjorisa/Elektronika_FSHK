Ky skedar përmban shënimet dhe kodin Python për llogaritjen e parametrave të qarqeve në laborator.

```python
# Llogaritja e frekuencës së prerjes për qarkun RC
import numpy as np

R = 1000.0  # Rezistenca në Ohm
C = 1e-7    # Kapaciteti në Farad

fc = 1 / (2 * np.pi * R * C)
print(f"Frekuenca e Prerjes (fc): {fc:.2f} Hz")
```
