## 1. Qëllimi i Eksperimentit
Ky eksperiment ka për qëllim njohjen praktike me përdorimin e pllakës së provës (breadboard) dhe ndërtimin e qarqeve elementare elektronike. Synohet kuptimi i lidhjeve elektrike të brendshme të breadboard-it, përdorimi i instrumenteve bazë të maturimit, si dhe verifikimi praktik i sjelljes së qarqeve me komponente diskrete dhe amplifikues operacionalë.

---

## 2. Aparatura dhe Pajisjet e Përdorura
Për realizimin e punës praktike u përdor stacioni eksperimental i montimit dhe testimit:
- **Pllaka e Provës (Breadboard):** Pllakë pa saldim për lidhjen e shpejtë të komponenteve. Përmban shinat anësore për ushqimin me energji ($+V$, $-V$, GND) dhe blloqet qendrore për ndërtimin e qarkut.
- **Gjeneratori i Funksioneve (Function Generator):** Për prodhimin e sinjaleve hyrëse periodike.
- **Oshiloskopi:** Për vëzhgimin vizual të sinjaleve në hyrje ($V_{in}$) dhe në dalje ($V_{out}$).
- **Ushqimi i Jashtëm DC (Power Supply):** Për furnizimin me tension simetrik të qarqeve të integruara (IC).

## 4. Formulat Themelore  
1. **Ligji i Ohm-it:**  
   $$V = I \cdot R$$
2. **Fitimi i Tensionit ($A_v$):**  
   $$A_v = \frac{V_{out}}{V_{in}}$$
3. **Fitimi në Decibel (dB):**  
   $$A_{dB} = 20 \cdot \log_{10}\left(\frac{V_{out}}{V_{in}}\right)$$
4. **Frekuenca e Prerjes (për qarqe RC / Filtra):**  
   $$f_c = \frac{1}{2 \pi R C}$$
