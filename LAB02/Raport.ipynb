## 1. Qellimi i Eksperimentit
Ky punim laboratorik ka për qëllim shqyrtimin dhe vërtetimin praktik të skemave të pjesëtimit të tensionit (voltage divider), si dhe vëzhgimin e sjelljes së qarqeve gjatë lidhjes së një rezistence ngarkese ($R_L$), duke bërë një krahasim midis të dhënave të regjistruara në eksperiment dhe parashikimeve teorike.

---

## 2. Te dhenat dhe Rezultatet e Eksperimenteve

### Eksperimenti 1: Pjestuesi Baze i Tensionit
Në këtë fazë u analizua qarku bazë i pjesëtimit të tensionit, i cili përbëhet nga dy rezistenca të vendosura në seri të furnizuara nga një burim tensioni hyrës.

* **Vlerat e matura dhe nominale:**
  - $V_{in} = 15,35\text{ V}$ 
  - $V_{out} = 7,5\text{ V}$ 
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
* **Formula e zbatimit:**
  $$\frac{V_{out}}{V_{in}} = \frac{R_2}{R_1 + R_2}$$
* Matjet dëshmojnë se rënia e tensionit në pikën e lidhjes ndërmjet rezistencave përputhet saktësisht me raportin përkatës të tyre.

---

### Eksperimenti 2: Qarku i Hapur (Pa Ngarkese)
U shqyrtua gjithashtu sjellja e qarkut me rezistenca të rangut të lartë (në shkallën e megaohmëve) kur ai ndodhet në skaj të hapur.

* **Parametrat e matur:**
  - $R_1 = 3,27\text{ M}\Omega$ 
  - $R_2 = 3,30\text{ M}\Omega$ 
  - $V_{in} = 6,82\text{ V}$ (ose $V = 15,77\text{ V}$ ne total te burimit) 
  - $V_{out} = 6,71\text{ V}$ 

---

### Eksperimenti 3: Qarku me Rezistence Ngarkese ($R_L$)
U vëzhgua se si ndikon integrimi i një rezistence ngarkese ($R_L$) në daljen e pjesëtimit të tensionit, ku kjo rezistencë lidhet paralelisht me rezistencën $R_2$.

* **Formula e ekuivalencës për rezistencat në paralel:**
  $$R_2 \parallel R_L = \frac{R_2 \cdot R_L}{R_2 + R_L}$$
* **Formula e tensionit ne dalje me ngarkese:**
  $$V_{out} = \frac{R_2 \parallel R_L}{R_1 + R_2 \parallel R_L} \cdot V_{in}$$

* **Rasti 1:**
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
  - $R_L = 3,30\text{ M}\Omega$ 
  - Tensioni i matur $V = 7,76\text{ V}$ 

* **Rasti 2:**
  - $R_1 = 7,55\text{ k}\Omega$ 
  - $R_2 = 7,78\text{ k}\Omega$ 
  - $R_L = 2,34\text{ M}\Omega$ 
  - Tensioni i matur $V = 7,71\text{ V}$

## 3. Perfundime
* Gjetjet praktike vërtetojnë shprehjet teorike të përdorura për pjesëtimin e tensionit.
* Futja e rezistencës së ngarkesës ($R_L$) modifikon rezistencën e përgjithshme ekuivalente të degës së poshtme ($R_2 \parallel R_L$), duke sjellë si pasojë një zvogëlim të tensionit në dalje ($V_{out}$) në krahasim me rastin kur qarku lihet pa ngarkesë.
* Të dhënat e marra në ambientin laboratorik përkojnë mirë me përllogaritjet teorike, duke llogaritur edhe devijimet e lejuara të komponentëve apo rezistencën e brendshme të vetë aparateve matëse.
